# FHE-Coder Demo: MLP over Encrypted Data (TFHE / C)

This notebook demonstrates the **FHE-Coder** agent (`B + Formal + Security Check + RAG` workflow) automatically generating a correct **C implementation** of a Multi-Layer Perceptron (MLP) using the [TFHE](https://tfhe.github.io/tfhe/) gate-bootstrapping library.

The agent uses a ReAct (Reasoning + Acting) loop with two tools:
- **`compile_execute_secure_code`** — saves the C code, compiles it with `gcc`, runs it, and validates it passes security checks
- **`rag`** — retrieves relevant TFHE API documentation via semantic search

It is guided by a **formal Dafny-like specification** of the MLP task and two reference C implementations.

## 1. Setup

```python
import os
import sys
import logging
import subprocess

# Ensure repo root is on sys.path
REPO_ROOT = os.path.abspath(".")
if REPO_ROOT not in sys.path:
    sys.path.insert(0, REPO_ROOT)

from dotenv import load_dotenv
load_dotenv()

assert os.getenv("OPENROUTER_API_KEY"), "OPENROUTER_API_KEY not set in .env"
assert os.getenv("OPENAI_API_KEY"),      "OPENAI_API_KEY not set in .env (needed for RAG embeddings)"
print("API keys loaded.")
```

## 2. Configure Logging and Output Directory

```python
RUN_ID = 0
MODEL  = "openai/gpt-5"
TASK   = "task_mlp"

save_dir = os.path.join(
    "logs_demo",
    MODEL.split("/")[1].replace(":", "_").replace("-", "_"),
    TASK,
    str(RUN_ID),
)
os.makedirs(save_dir, exist_ok=True)

log_file = os.path.join(save_dir, "run.log")
logger = logging.getLogger("fhe_coder_demo")
logger.setLevel(logging.INFO)
if logger.hasHandlers():
    logger.handlers.clear()
fh = logging.FileHandler(log_file)
fh.setFormatter(logging.Formatter("%(asctime)s - %(levelname)s - %(message)s"))
ch = logging.StreamHandler(sys.stdout)
ch.setFormatter(logging.Formatter("%(asctime)s - %(levelname)s - %(message)s"))
logger.addHandler(fh)
logger.addHandler(ch)

logger.info(f"Save dir : {save_dir}")
logger.info(f"Model    : {MODEL} | Task: {TASK} | Run: {RUN_ID}")
```

## 3. Define the Compile-Execute-Secure Tool

Unlike the Python/OpenFHE tasks, the MLP uses **TFHE C code**.
The tool therefore:
1. Saves the generated code as `program.c`
2. Compiles it with `g++` against `-I/usr/local/include -L/usr/local/lib -ltfhe-spqlios-fma`
3. Runs the resulting binary
4. Runs the TFHE security check (`validate_tfhe_program` + `minimum_lambda` extraction)

```python
from agents.tool import tool
from tools.security_check import check_secure
from tools.summary_rag import SummaryRAG

C_FILE   = os.path.join(save_dir, "program.c")
EXE_FILE = os.path.join(save_dir, "program")


def _save_c_code(code: str) -> None:
    with open(C_FILE, "w") as f:
        f.write(code)


def _compile_c() -> tuple[bool, str]:
    """Compile program.c → program using g++ with TFHE libs. Returns (success, message)."""
    try:
        result = subprocess.run(
            ["g++", C_FILE, "-o", EXE_FILE,
             "-I/usr/local/include", "-L/usr/local/lib", "-ltfhe-spqlios-fma"],
            capture_output=True,
            text=True,
            check=False,
        )
    except Exception as e:
        return False, f"Error during compilation: {e}"
    if result.returncode == 0:
        return True, "Compilation successful."
    return False, f"Compilation failed: {result.stderr[:150]}"


def _run_binary() -> tuple[bool, str]:
    """Run compiled binary. Returns (success, stdout/stderr)."""
    try:
        result = subprocess.run(
            [EXE_FILE],
            capture_output=True,
            text=True,
            timeout=120,
        )
        if result.returncode == 0:
            return True, result.stdout.strip()
        return False, f"Runtime error (code {result.returncode}):\n{result.stderr[:500]}"
    except subprocess.TimeoutExpired:
        return False, "Execution timed out (>120 s)."


@tool
def compile_execute_secure_code(code: str) -> str:
    """
    Saves the given C code, compiles it with g++ against the TFHE library,
    runs the binary, and validates TFHE security requirements.

    Args:
        code (str): The C source code to compile and run.

    Returns:
        str: Success message with program output, or a detailed error message.
    """
    _save_c_code(code)

    compiled, compile_msg = _compile_c()
    if not compiled:
        return compile_msg

    ran, run_output = _run_binary()
    if not ran:
        return run_output

    security_result = check_secure(C_FILE)
    if security_result != "Code is secure!":
        return f"Execution succeeded but security check failed: {security_result}"

    return f"Passed all tests. Program output:\n{run_output}"


# ---- RAG tool ----

retriever = SummaryRAG(
    summary_db_path="./tfhe_documentation/summaries_db.json",
    embedding_model="text-embedding-3-small",
    persist_directory="./chroma_tfhe_summaries",
)


@tool
def rag(query: str) -> str:
    """
    Retrieves the most relevant TFHE function signatures and Doxygen documentation for the query.

    Args:
        query (str): The query describing the TFHE operation you need.

    Returns:
        str: Retrieved function signature and Doxygen documentation.
    """
    sig, summary, doxygen = retriever.retrieve(query, k=1)[0]
    return "\n".join([doxygen, sig])


print("Tools ready: compile_execute_secure_code, rag")
```

```python
from agents.react_agent import ReactAgent

agent = ReactAgent(
    tools=[compile_execute_secure_code, rag],
    model=MODEL,
    seed=RUN_ID,
    logger=logger,
)

print(f"Agent initialised. Model: {MODEL}")
print("Starting agent run...\n" + "=" * 60)
```

## 4. Task Prompt

The agent is given a formal Dafny-like specification for the MLP task — security parameter, network architecture, required TFHE building blocks, and code structure.

```python
task_mlp_formal_prompt = """
You are an expert coding agent familiar with cryptographic concepts.

Make sure to follow the below instructions:

    - The input values (weights, biases, activations) are hardcoded in the program.

    - Code should not have any extra logging or print statements except the final result.

    - Code must not be a basic C program — it must use TFHE library functions for all arithmetic.

    - The required TFHE headers are:
            #include <tfhe/tfhe.h>
            #include <tfhe/tfhe_io.h>

    - Code should have `assert` statements to implement the `ensures` and `requires` clauses.

    - Follow the following Dafny-like code as a pseudo-code specification:

// Global security parameter for TFHE gate bootstrapping
const minimum_lambda: int := 110;

method Main(input_dim: int, hidden_dim: int, output_dim: int)
  returns (result_shape: int)
  requires input_dim > 0
  requires hidden_dim > 0
  requires output_dim > 0
  ensures result_shape == output_dim
{
  var Z1: encrypted_matrix := fhe_matmul(X, W1);
  var H1: encrypted_matrix := fhe_relu(Z1);
  var H1_32: encrypted_matrix := fhe_truncate(H1);
  assert H1_32.shape == (M, hidden_dim);

  var Y: encrypted_matrix := fhe_matmul(H1_32, W2);
  assert Y.shape == (M, output_dim);

  result_shape := output_dim;
  assert result_shape == output_dim;
}

    - Follow these informal code requirements strictly:

### Multi-Layer Perceptron (MLP) over TFHE-Encrypted Data

**Security Context:**
* Use `new_default_gate_bootstrapping_parameters(minimum_lambda)` with `minimum_lambda = 110`.
* Use `new_random_gate_bootstrapping_secret_keyset` for key generation.
* All arithmetic must use the cloud key (`bk->cloud`).

**Network Architecture (hardcoded values):**
* Input X (3x2):  rows = [1,2], [-3,4], [0,-1]
* Layer 1 weight W1 (2x2): [[2,-1],[3,1]]
* Layer 1: Z1 = X @ W1, then ReLU, then truncate 64->32 bit -> H1
* Layer 2 weight W2 (2x1): [[4],[-2]]
* Layer 2: Y = H1 @ W2, linear (no activation)
* Encrypt all values using `bootsCONSTANT` (bit-by-bit, LSB-first).

**Building blocks required:**
* full_adder, fhe_add_bits, fhe_mult_unsigned, fhe_conditional_negate,
  fhe_mul_signed32x32_to64, fhe_relu_bits, fhe_matmul_32x32_to_64,
  fhe_truncate64_to32_matrix

**Output:** print 3 decrypted int64 values, one per line. Full cleanup at end.

TASK: Write a C program using the TFHE library that implements a 2-layer MLP on
encrypted inputs and prints the decrypted output.
""".strip()

user_prompt = task_mlp_formal_prompt
print(f"Prompt ready. ({len(user_prompt):,} chars)")
```

```python
output = agent.run(user_prompt, max_rounds=10)
print("\n" + "=" * 60)
print("Agent finished.")
```

## 8. Inspect the Generated C Code

```python
if os.path.exists(C_FILE):
    with open(C_FILE, "r") as f:
        generated_code = f.read()
    print(f"Generated code saved to: {C_FILE}")
    print("\n" + "=" * 60 + "\n")
    print(generated_code)
else:
    print("No C file found — agent may not have called compile_execute_secure_code.")
    print("Agent final response:\n", output)
```

## 9. Run the Generated MLP Binary

Recompile (if needed) and run the final binary to observe the decrypted MLP output.

```python
if os.path.exists(C_FILE):
    # Recompile to be sure
    ok, msg = _compile_c()
    print(msg)

    if ok:
        ok2, out2 = _run_binary()
        print("\nProgram output:")
        print(out2 if ok2 else f"Run error: {out2}")
else:
    print("No C file to compile.")
```

## 10. Security Check

```python
if os.path.exists(C_FILE):
    result = check_secure(C_FILE)
    print(f"Security check result: {result}")
else:
    print("No C file to check.")
```
